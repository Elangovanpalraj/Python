```python
def factorial(n):
    if n==1:
        return 1
    else:
        return n*factorial(n-1)
a=5
#print(f"factorial of {number} is {factorial(number)}")
print(factorial(a))
```

```python
names = ['Saran', 'Santhosh', 'Arun', 'Ameer', 'Vicky']

s_name = None
a_name = None
for name in names:
    if name.startswith('S') and s_name is None:
        s_name = name
    elif name.startswith('A') and a_name is None:
        a_name = name
    if s_name and a_name:
        break

filtered_names = [name for name in [s_name, a_name] if name]

print(filtered_names)
```
