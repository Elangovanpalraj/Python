```python
class ExampleClass:
    def __init__(self, data,divisor):
        self.data = data
        self.divisor=divisor

    def divide(self):
        try:
            result = self.data / self.divisor
        except :
            result = None
        else:
            print("Division successful!")
        finally:
            print("Execution completed.")
        
        return result
data=int(input("enter a num"))
divisor=int(input("enter a num2"))
example = ExampleClass(data,divisor)
print(example.divide())

```
