```python
sum_even=0
num=2
count=0
while count<10:
    sum_even+=num
    num+=2
    count+=1
print(sum_even)
    
```

```python
a,b=0,1
count=0
n=10
print("fabnaocci series")
while count<n:
    print(a,end=" ")
    a,b=b,a+b
    count+=1
```
