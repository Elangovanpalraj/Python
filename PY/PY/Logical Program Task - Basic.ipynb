```python
#33 Count of all caps small digits special count without special methods?
word="6The 9quick 4brown 2fox 0jumps over the lazy dog@@@@#####"
print("the given string is \n"+word)
lower=0
upper=0
no=0
special=0
sp=0
for i in word:
    if 'a'<=i<='z':
        lower+=1
    elif 'A'<=i<='Z':
        upper+=1
    elif '0'<=i<='9':
        no+=1
    elif i==" ":
        sp+=1
    else:
        spl+=1
print(f"\n the number of lowercase is {lower} \n the number of capital is {upper} \n the no of digits is {no} \n and the no of special characters is {special}")
```

```python
#34 Count of all caps small digits special count with special methods?
word="6The 9quick 4brown 2fox 0jumps over the lazy dog@@@@#####"
print("the given string is \n"+word)
alpcaps=0
alpsml=0
dgs=0
sp=0
spcl=0
for i in word:
    if i.isalpha():
        if i.isupper():
            alpcaps+=1
        elif i.islower():
            alpsml+=1
    elif i.isdigit():
        dgs+=1
    elif i==" ":
        sp+=1
    else:
        spl+=1
print(f"\n the number of lowercase is {alpsml} \n the number of capital is {alpcaps} \n the no of digits is {dgs} \n and the no of special characters is {spcl}")
```

```python
#43 Increase the current date by 10?
import datetime as dt
today=dt.date.today()
print(f"today date is {today}")
x=dt.timedelta(days=10)
print(f"date after ten days is {today+x}")
```

```python
#44 Increase the current month by 2?
import datetime as dt
from dateutil.relativedelta import relativedelta as rel
today=dt.date.today()
x = rel(months=2)  
print(f"today date is {today}")
print(f"month after two months date is {today+x}")
```

```python
#45 Increase the current year by 3?
import datetime as dt
from dateutil.relativedelta import relativedelta as rel
today=dt.date.today()
x = rel(years=3)  
print(f"today date is {today}")
print(f"year after three years date is {today+x}")
```
