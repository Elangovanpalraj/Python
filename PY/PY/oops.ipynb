```python
#Encapsulation
class HotelRoom:
    def __init__(self, room_number, room_type, rate_per_night):
        self.__room_number = room_number  
        self.__room_type = room_type        
        self.__rate_per_night = rate_per_night  
        self.__is_available = True       

    
    def get_room_number(self):
        return self.__room_number

    def get_room_type(self):
        return self.__room_type

    def get_rate_per_night(self):
        return self.__rate_per_night

    def is_available(self):
        return self.__is_available

    
    def set_availability(self, is_available):
        self.__is_available = is_available

    
    def display_info(self):
        print(f"Room Number: {self.__room_number}")
        print(f"Room Type: {self.__room_type}")
        print(f"Rate per Night: ${self.__rate_per_night}")
        print(f"Available: {'Yes' if self.__is_available else 'No'}")



room = HotelRoom(14, "Deluxe", 10000)

print("Room Information:")
room.display_info()


room.set_availability(False)

print("Updated Room Information:")
room.display_info()
```

```python
#Abstraction
class Area_Perimeter():
    def calculate_area(self):
        pass
    def calculate_perimeter(self):
        pass
class Rectangle(Area_Perimeter):
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth
    
    def calculate_area(self):
        return self.length * self.breadth
    
    def calculate_perimeter(self):
        return 2 * (self.length + self.breadth)

class Circle(Area_Perimeter):
    def __init__(self, radius):
        self.radius = radius
    
    def calculate_area(self):
        return 3.14 * self.radius ** 2
    
    def calculate_perimeter(self):
        return 2 * 3.14 * self.radius
rectangle = Rectangle(5, 10)
print(f"Rectangle = Area: {rectangle.calculate_area()}, Perimeter: {rectangle.calculate_perimeter()}")

circle = Circle(4)
print(f"Circle = Area: {circle.calculate_area()}, Perimeter: {circle.calculate_perimeter()}")
```
