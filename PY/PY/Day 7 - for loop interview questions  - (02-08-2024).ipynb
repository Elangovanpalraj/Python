```python
#interview question
#print even sum form 1 to 90
count=0
for i in range(1,91):
    if i%2==0:
        print("even no is:",i)
        count+=i
print("sum is:",count)
```

```python
#print odd sum form 1 to 90
count=0
for i in range(1,91):
    if i%2!=0:
        print("odd no is:",i)
        count+=i
print("sum is",count)
```

```python
#print sum of even or odd form 1 to 90
count=0
for i in range(1,91):
    if i%2==0:
        print("even no is:",i)
        count+=i
        print("sum is",count)
    else:
        print("odd no is:",i)
        count+=i
        print("sum is",count)
```
