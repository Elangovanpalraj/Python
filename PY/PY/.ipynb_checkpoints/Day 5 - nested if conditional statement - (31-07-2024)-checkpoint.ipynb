```python
#school
maths=int(input("enter maths marks"))
science=int(input("enter science marks"))
social=int(input("enter social marks"))
tamil=int(input("enter tamil marks"))
english=int(input("enter english marks"))
avg=89

if ((maths + science + social + tamil + english)/5)>=90:
    avg="first class"
elif ((maths + science + social + tamil + english)/5)<=89 and ((maths + science + social + tamil + english)/5)>=80:
    avg="second class"
if (maths+science and social and tamil and english)>=35:
    print("all pass",avg)
else:
    print("fail")
```
