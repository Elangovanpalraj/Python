```python
def even(a):
  evenn=a%2==0
  return evenn
a=range(1,10)
list(filter(even,a))
```

```python
#in both map and filter there will be two arguments, first one is the user defined function or lambda function( which will act as operator)
# and the second one is the list in which the operation (lambda or user def func) should be done
#in filter and map, the user defined function or the lambda is necessary
```

```python
x=list(range(1,10))
sq= lambda x:x**2
print(list(map(sq,x)))
y=list(range(1,50))
issq=lambda x:x**0.5==int(x**0.5)
print(list(filter(issq,y)))
```

```python
def isprime(n):
  if n<=0:
    return False
  if n==1:
    return False
  li=range(1,n+2)
  prim=[]
  for i in range(1,n+1):
    if n%i==0:
      prim.append(i)
  if len(prim)==2:
    return True
  else:
    return False
x=list(range(1,100))
print(list(filter(isprime,x)))
def double(a):
  return a*2
y=list(range(1,15))
print(list(map(double,y)))
```

```python
from functools import reduce
a=list(range(1,10))
addi=lambda x,y:x+y
print(reduce(addi,a))
```

```python
multi=lambda x,y:x*y
print(reduce(multi,list(range(1,6))))
```

```python
 #odd no from 1 to 10
 isodd=lambda x: x%2!=0
 print(list(filter(isodd,list(range(1,10)))))
```

```python
iseven=lambda x:x%2==0
print(list(filter(iseven,list(range(1,60)))))
```

```python
#even sum 1 to 90
from functools import reduce
add=lambda x,y:x+y
li=list(filter(iseven,list(range(1,91))))
print(reduce(add,li))
```

```python
li2=list(filter(isodd,list(range(1,81))))
print(reduce(add,li2))
```

```python
from functools import reduce
def fa(n):
  return reduce(lambda x,y:x*y,list(range(1,n+1)))
print(fa(5))
```
