```python
#17. print all prime nos from 1 to 50
def is_primeno(n):
    la=[]
    if n==1:
        return False
    for i in range(2,n):
        if n%i==0:
            la.append(i)
    if len(la)==0:
        return True
    else:
        return False
for i in range(1,51):
    if is_primeno(i):
        print(i)
```

```python
#33.Count of all caps small digits special count without special methods
ta="6The 9quick 4brown 2fox 0jumps over the lazy dog@@@@#####"
print("the given string is \n"+ta)
lwr=0
upr=0
no=0
spl=0
spc=0
for i in ta:
    if 'a'<=i<='z':
        lwr+=1
    elif 'A'<=i<='Z':
        upr+=1
    elif '0'<=i<='9':
        no+=1
    elif i==" ":
        spc+=1
    else:
        spl+=1
print(f"\n the number of lowercase is {lwr} \n the number of capital is {upr} \n the no of digits is {no} \n and the no of special characters is {spl}")

```

```python
#34.Count of all caps small digits special count with special methods?
ta="6The 9quick 4brown 2fox 0jumps over the lazy dog@@@@#####"
print("the given string is \n"+ta)
alcap=0
alsmo=0
dig=0
spc=0
spl=0
for i in ta:
    if i.isalpha():
        if i.isupper():
            alcap+=1
        elif i.islower():
            alsmo+=1
    elif i.isdigit():
        dig+=1
    elif i==" ":
        spc+=1
    else:
        spl+=1
print(f"\n the number of lowercase is {alsmo} \n the number of capital is {alcap} \n the no of digits is {dig} \n and the no of special characters is {spl}")
```

```python
#40. print duplicates in an array
arr=[1,2,3,4,4,5,8,9,7,8,3]

seen = []
duplicates = []

for num in arr:
    if num in seen:
      duplicates.append(num)
    else:
      seen.append(num)

if duplicates:
    print("Duplicate elements:", duplicates)
else:
    print("No duplicates found.")
```
