```python
def count_characters(input_string):
    upper_count =0
    lower_count=0
    special_count=0
    for char in input_string:
        if char.isupper():
            upper_count+=1
        elif char.islower():
            lower_count+=1
        else:
            special_count+=1
    return upper_count,lower_count,special_count
input_string="Virat Kohli 18"
upper,lower,special=count_characters(input_string)
print(upper)
print(lower)
print(special)
```

```python
user=input("enter a captcha:")
caps_count=0
small_count=0
special_count=0
for i in user:
    if "A"<=i<="Z":
        caps_count+=1
    elif "a"<=i<="z":
        small_count+=1
    else:
        special_count+=1
print(caps_count)
print(small_count)
print(special_count)
```

```python
def reverse_each_word(input_string):
    words = input_string.split()
    reversed_words = [i[::-1] for i in words]  #
    reversed_string = ' '.join(reversed_words)
    return reversed_string
input_str = "Hello, i am from chengalpattu"
reversed_str = reverse_each_word(input_str)
print(reversed_str)
```

```python
words= ['zero','one','two','three','four','five','six','seven','eight','nine']
def number_to_word(n):
    if(n==0):
        return ""
    else:
         result= words[n%10]
         results = number_to_word(int(n/10)) + result + " "
    return results
n = int(input("enter a number:"))
print("you entered the num : ",n)
print("becomes it words: ",end="")
print(number_to_word(n))
```
